```python
import pandas as pd
```

```python
d=pd.read_csv("C:\\Users\\steve\\Downloads\\titanic\\Titanic-Dataset.csv")

```

```python
df=pd.DataFrame(d)
```

```python
df
```

```python
df.isnull()
```

```python
df.dropna()
```

```python
def survive(Survived):
    if Survived==1:
        return "Yes"
    else:
        return "NO"
df['survive']=df['Survived'].apply(survive) 
print(df)
```

```python
df
```

```python
df.drop('Survived',axis=1,inplace=True)
```

```python
df
```

```python
def Ticket_Class(Pclass):
    if Pclass==1:
        return "First Class"
    elif Pclass==2:
        return "second Class"
    else:
        return "Third Class"
df['Ticket_Class']=df['Pclass'].apply(Ticket_Class)
print(df)
         
```

```python
df
```

```python
df.drop('Pclass',axis=1,inplace=True)
```

```python
df
```

```python
df.drop('Cabin',axis=1,inplace=True)
```

```python
df
```

```python
df.dropna(subset=['Age'],inplace=True)
```

```python
df
```

```python
df.dropna(subset=['Embarked'],inplace=True)
```

```python
df
```

```python
def passanger_boardship(Embarked):
    if Embarked=='S':
        return  "Southampton (England)"
    elif Embarked=='C':
        return "Cherbourg (France)"
    else:
        return " Queenstown (now Cobh, Ireland)"
df['passanger_boardship']=df['Embarked'].apply(passanger_boardship)
print(df)
    
```

```python
df
```

```python
df.drop('Embarked',axis=1,inplace=True)
```

```python
df
```

```python
df.describe()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3]
y = [100, 200, 150]

plt.plot(x, y)
plt.title('Passenger Counts')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.xticks([1, 2, 3], ['First', 'Second', 'Third']) 
plt.yticks([0, 100, 200])
plt.show()
```

```python
survived_counts = df['survive'].value_counts()
labels = ['Not Survived', 'Survived']
colors = ['lightcoral', 'lightgreen']
plt.figure(figsize=(6, 6))
plt.pie(survived_counts, labels=labels, autopct='%1.1f%%', startangle=90, colors=colors)
plt.title('Survival Distribution on Titanic')
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt

pclass_counts = df['Ticket_Class'].value_counts().sort_index()

plt.figure(figsize=(6, 4))
plt.bar(pclass_counts.index.astype(str), pclass_counts.values, color='skyblue')
plt.title('Passenger Count by Class (Bar Chart)')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(df['Age'], df['Fare'], alpha=0.5, color='purple')
plt.title('Scatter Plot: Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True)
plt.show()
```

```python
embarked_counts = df['passanger_boardship'].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(embarked_counts.index, embarked_counts.values, color='orange')
plt.title('Column Chart: Passengers by Embarkation Port')
plt.xlabel('Port')
plt.ylabel('Number of Passengers')
plt.show()
```
